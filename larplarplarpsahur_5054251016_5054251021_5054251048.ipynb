# 6. Distribusi Target

```python
sns.countplot(x='target', data=df)
plt.title('Distribusi Target')
plt.show()
```

```python
# Hitung matriks korelasi Pearson
corr_matrix = df.corr()

# Tampilkan tabel korelasi (buat Tabel di makalah)
display(corr_matrix)

# Visualisasi Heatmap
plt.figure(figsize=(12, 9))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Matriks Korelasi Antar Variabel (Heatmap)')
plt.tight_layout()
plt.savefig('heatmap_korelasi.png', dpi=300)
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Boxplot thalach
sns.boxplot(
    ax=axes[0],
    data=df,
    x="target",
    y="thalach",
    hue="target",
    palette="Pastel1",
    legend=False,
)
axes[0].set_title("Distribusi thalach berdasarkan Target")
axes[0].set_xticks([0, 1])
axes[0].set_xticklabels(["Tidak Sakit (0)", "Sakit (1)"])

# Boxplot oldpeak
sns.boxplot(
    ax=axes[1],
    data=df,
    x="target",
    y="oldpeak",
    hue="target",
    palette="Pastel2",
    legend=False,
)
axes[1].set_title("Distribusi oldpeak (ST Depression) berdasarkan Target")
axes[1].set_xticks([0, 1])
axes[1].set_xticklabels(["Tidak Sakit (0)", "Sakit (1)"])

plt.tight_layout()
plt.savefig("boxplot_thalach_oldpeak.png", dpi=300)
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Hubungan Sex dengan Target
sns.countplot(ax=axes[0], data=df, x="sex", hue="target", palette="Set2")
axes[0].set_title("Proporsi Target berdasarkan Jenis Kelamin")
axes[0].set_xticks([0, 1])
axes[0].set_xticklabels(["Wanita (0)", "Pria (1)"])

# Hubungan Chest Pain Type (cp) dengan Target
sns.countplot(ax=axes[1], data=df, x="cp", hue="target", palette="Set2")
axes[1].set_title("Proporsi Target berdasarkan Tipe Nyeri Dada")
axes[1].set_xlabel("Chest Pain Type (0 s/d 3)")

# Hubungan Exercise Induced Angina (exang) dengan Target
sns.countplot(ax=axes[2], data=df, x="exang", hue="target", palette="Set2")
axes[2].set_title("Proporsi Target berdasarkan Exang")
axes[2].set_xticks([0, 1])
axes[2].set_xticklabels(["Tidak (0)", "Ya (1)"])

plt.tight_layout()
plt.savefig("countplot_kategorik.png", dpi=300)
plt.show()
```

```python
selected_features = ['age', 'trestbps', 'chol', 'thalach', 'oldpeak', 'target']
sns.pairplot(df[selected_features], hue='target', palette='bwr', diag_kind='kde')
plt.suptitle('Pairplot Fitur Numerik Utama terhadap Target', y=1.02)
plt.savefig('pairplot_selected.png', dpi=300)
plt.show()
```

```python
num_cols = ['age', 'trestbps', 'chol', 'thalach', 'oldpeak']

# Hitung Skewness
print("Nilai Skewness Fitur Numerik:")
print(df[num_cols].skew())

# Plot Distribusi (Histogram + KDE)
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
axes = axes.flatten()

for i, col in enumerate(num_cols):
    sns.histplot(data=df, x=col, kde=True, ax=axes[i], color='royalblue')
    axes[i].set_title(f'Distribusi {col} (Skew: {df[col].skew():.2f})')

# Hilangkan subplot kosong keenam
fig.delaxes(axes[5])
plt.tight_layout()
plt.show()
```

```python
outlier_data = []

for col in num_cols:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    outlier_data.append({
        'Fitur': col,
        'Q1': q1,
        'Q3': q3,
        'IQR': iqr,
        'Batas Bawah': lower_bound,
        'Batas Atas': upper_bound,
        'Jumlah Outlier': len(outliers),
        'Persentase (%)': round((len(outliers) / len(df)) * 100, 2)
    })

df_outlier = pd.DataFrame(outlier_data)
display(df_outlier)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Crosstab cp vs target (persentase baris)
ct_cp = pd.crosstab(df['cp'], df['target'], normalize='index') * 100
sns.heatmap(ct_cp, annot=True, fmt='.1f', cmap='Blues', cbar=False, ax=axes[0])
axes[0].set_title('Persentase Diagnosis per Tipe Nyeri Dada (cp) [%]')
axes[0].set_xlabel('Target (0: Sehat, 1: Sakit)')
axes[0].set_ylabel('Tipe Nyeri Dada (0-3)')

# Crosstab sex vs target (persentase baris)
ct_sex = pd.crosstab(df['sex'], df['target'], normalize='index') * 100
sns.heatmap(ct_sex, annot=True, fmt='.1f', cmap='Blues', cbar=False, ax=axes[1])
axes[1].set_title('Persentase Diagnosis per Jenis Kelamin [%]')
axes[1].set_xticklabels(['Sehat (0)', 'Sakit (1)'])
axes[1].set_yticklabels(['Wanita (0)', 'Pria (1)'])

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x='age',
    y='thalach',
    hue='target',
    style='cp',
    palette='bwr',
    alpha=0.8,
    s=70
)
plt.title('Interaksi Usia, Thalach, dan Chest Pain Type terhadap Target')
plt.xlabel('Usia')
plt.ylabel('Detak Jantung Maksimum (thalach)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
# Korelasi mutlak terhadap target
target_corr = df.corr()['target'].drop('target').abs().sort_values(ascending=False)

plt.figure(figsize=(9, 5))
sns.barplot(x=target_corr.values, y=target_corr.index, hue=target_corr.index, palette='magma', legend=False)
plt.title('Peringkat Relevansi Fitur terhadap Diagnosis (Absolute Correlation)')
plt.xlabel('|Korelasi Pearson terhadap Target|')
plt.ylabel('Fitur')
plt.tight_layout()
plt.show()
```
